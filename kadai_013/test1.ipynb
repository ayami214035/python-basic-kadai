```python
import random
# 指定された数値のリストからランダムに1つ選択します
var = random.choice([0, 3, 5, 15, 16, 20, 21])
print(var)

# 3と5の両方の倍数であるかを最初に判定します
if var % 3 == 0 and var % 5 == 0:
  print("FizzBuzz")
elif var % 3 == 0:
  print("Fizz")
elif var % 5 == 0:
  print("Buzz")
else:
  print(var)
```

```python
user_names = ["侍太郎", "侍一郎", "侍二郎", "侍三郎", "侍四郎"]

# 検索したいユーザー名を代入する変数
target = "侍二郎"

for user_name in user_names:
    print(user_name)

    # 変数user_nameと変数targetの値が一致すれば、break文で繰り返し処理を強制終了する
    if (user_name == target):
        print(f"{target}さんが見つかったので、繰り返し処理を強制終了します。")
        break
```

```python
score = {
    "国語": 80,
    "数学": 55,
    "理科": 70,
    "社会": 85,
    "英語": 60
}

print("合格した科目は以下のとおりです。")

for key, value in score.items():
    # 変数valueの値（点数）が70より小さければ、キー（科目名）と値（点数）を出力せずにcontinue文で次のループに進む
    if (value < 70):
        continue
    print(f"{key} : {value}点")
```

```python
array = ["水","金","地","火","木","土","天","海","冥"]

for i in array:
  print(i)

# 9回まで繰り返す条件式を入れる
count = 0

while count < 9:
  print(array[count])
  count = count + 1
```
